```python
nums=[1,2,-1,3,0,4]
```

## Reversing Array Elements

```python
n=len(nums)
for i in range(n//2):
    tmp=nums[i]
    nums[i]=nums[n-i-1]
    nums[n-i-1]=tmp
print(nums)    
```

```python
nums[:]=nums[::-1]
print(nums)
```

## First Missing Number

```python
nums=[1,2,3,4,6,7,8]
n=len(nums)+1

print("Missing number is:",((n*(n+1))/2)-(sum(nums)))
```

```python
freq={}

for num in [1,2,3,4,4,3,3]:
    if num not in freq:
        freq[num]=1
    else:
        freq[num]+=1
print(freq)        
```

```python
nums = [2, 8, 11, 7, 15]
target = 9

for i,num in enumerate(nums):
    if target-num in nums:
        print(i,num)
```

```python
seen = {}

for i, num in enumerate(nums):
    complement = target - num

    if complement in seen:
        print([seen[complement], i])

    seen[num] = i
    print(seen[num],end=" ")
```
