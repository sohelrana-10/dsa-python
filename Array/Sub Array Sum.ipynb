```python
nums = [-2,1,-3,4,-1,2,1,-5,4]

maxi = float("-inf")

for i in range(len(nums)):
    curr_sum = 0
    for j in range(i, len(nums)):
        curr_sum += nums[j]
        maxi = max(maxi, curr_sum)

print(maxi)
```
